```python
import BookTupDB
```

```python
user_choice = """
Enter character:
'a' - Add a new book
'l' - List all books
's' - Search for a book
'd' - Delete a book
'p' - Update book price
'q' - Quit
Enter your choice: 
"""
```

```python
def prompt_add_book():
    name = input("Enter book name: ")
    author = input("Enter author name: ")
    price = float(input("Enter price of the book: "))
    year = int(input("Enter published year: "))
    genre = input("Enter book genre: ")

    BookTupDB.insert_book(name, author, price, year, genre)
```

```python
def list_books():
    books = BookTupDB.get_all_books()
    if not books:
        print("No books available")
        return
    for book in books:
        print(
            f"{book[0]} written by {book[1]} "
            f"|| Price: {book[2]} "
            f"|| Year: {book[3]} "
            f"|| Genre: {book[4]}"
        )
```

```python
def prompt_search_book():
    name = input("Enter book name: ")
    book = BookTupDB.search_book(name)
    if book:
        print(f"{book[0]} written by {book[1]}")
        print(f"Price: {book[2]}")
        print(f"Year: {book[3]}")
        print(f"Genre: {book[4]}")
    else:
        print("Book not found")
```

```python
def prompt_delete_book():
    name = input("Enter book name: ")
    BookTupDB.delete_book(name)
```

```python
def prompt_update_price():
    name = input("Enter book name: ")
    new_price = float(input("Enter new price: "))
    BookTupDB.update_price(name, new_price)
```

```python
user_input = input(user_choice)

while user_input.lower() != 'q':
    if user_input == 'a':
        prompt_add_book()
    elif user_input == 'l':
        list_books()
    elif user_input == 's':
        prompt_search_book()
    elif user_input == 'd':
        prompt_delete_book()
    elif user_input == 'p':
        prompt_update_price()
    else:
        print("Invalid choice!")
    user_input = input(user_choice)
```
