```python
import DB
```

```python
user_choice= """ Enter character: 
- 'a' to add new book into DB
- 'l' to list all books from DB
- 's' to search for a book in DB
- 'd' to delete a book from DB
- 'p' to update book price
- 'q' to quit/exit
Enter your choice:
"""
# print(user_choice)
```

```python
def prompt_add_book():
    name=input("Enter book name: ")
    author=input("Enter author name: ")
    price=float(input("Enter price of the book: "))
    year=int(input("Enter published year: "))
    genre=input("Enter book genre: ")
    DB.insert_book(name,author,price,year,genre)
```

```python
def list_books():
    books=DB.get_all_books()
    for book in books:
        print(f"{book['Name']} written by {book['Author']}"
             f"|| Year: {book['Year']}"
              f"|| Genre: {book['Genre']}"
              f"|| Price: {book['Price']}")
```

```python
def prompt_search_book():
    name=input("Enter Book name: ")
    book=DB.search_book(name)
    if book:
        print("Book Avaiable")
        print(f"{book['Name']} written by {book['Author']}"
             f"|| Year: {book['Year']}"
              f"|| Genre: {book['Genre']}"
              f"|| Price: {book['Price']}")
    else:
        print("Book not found")
```

```python
def prompt_delete_book():
    name=input("Enter Book name to delete: ")
    DB.delete_book(name)
```

```python
def prompt_update_price():
    name=input("Enter Book name: ")
    new_price=float(input("Enter new price: "))
    DB.update_price(name,new_price)
```

```python
user_input=input(user_choice)
while user_input !='q':
    if user_input == 'a':
        prompt_add_book()
    elif user_input == 'l':
        list_books()
    elif user_input == 's':
        prompt_search_book()
    elif user_input == 'd':
        prompt_delete_book()
    elif user_input == 'p':
        prompt_update_price()
    else:
        print("Invalid Choice! ")
    user_input = input(user_choice)
```
