```python
# Write a Python program that determines if a given number is a perfect square.
num = int(input("Enter a number:"))
i=1
while (i*i<=num):
    if ( i*i==num):
        print("Perfect Square")
        break
    i+=1
else:
    print("Not a Perfect Square")
```

```python
# Write a Python program that determines if a given number is a perfect square.
import math
num = int(input("Enter a number:"))
root=int(math.sqrt(num))
if ( root*root == num):
    print("Perfect Square")
else:
    print("Not a perfect square")
```
