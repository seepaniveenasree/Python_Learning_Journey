Python function which accepts a list of strings containing details of deposits and withdrawals made in a bank account and returns the net amount in the account. 

Suppose the following input is supplied to the function ["D:300","D:300","W:200","D:100"] where D means deposit and W means withdrawal, then the net amount in the account is 500.


```python
def dw_amt(li):
    d=w=0
    for i in li:
        sl=i.split(":")
        if(sl[0]=='D'):
            d=d+int(sl[1])
        else:
            w=w+int(sl[1])
    res=d-w
    return res
li=["D:300","D:300","W:200","D:100"]
amt=dw_amt(li)
amt
```

```python
#  Finding all occurance of a substring in an inputstring
def sub_count(sstr,sub_sstr):
    cnt=0
    lst=[]
    found=sstr.find(sub_sstr)
    while(found!=-1):
        lst.append(found)
        cnt=cnt+1
        found=sstr.find(sub_sstr,found+1)
    return cnt,lst
sstr=input().strip()
sub_sstr=input().strip()
cnt,lst=sub_count(sstr,sub_sstr)
print(cnt,lst)
```

```python
s = 'year 2026 @home!'
letters = digits = spaces = specials = 0

for c in s:
    if c.isalpha():
        letters += 1
    elif c.isdigit():
        digits += 1
    elif c.isspace():
        spaces += 1
    else:
        specials += 1

print("Letters:", letters)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", specials)
```

```python
###  program to print even length words in a string

def even_len(s1):
    lis1=s1.split()
    lsout=[]
    for i in lis1:
        l=len(i)
        if (l%2==0):
            lsout.append(i)
    return lsout


s1="Python programming is a high level language"
res=even_len(s1)
print(res)
```

```python
# An anagram is a word or phrase formed by rearranging the letters of a different words.
def anagram(s1,s2):
    if(sorted(s1)==sorted(s2)):
        return "Is Anagram"
    else:
        return "Not a Anagram"
s1="act"
s2="cat"
print(anagram(s1,s2))
```

```python
##### Python function that takes a list of words and return the longest word and the length of the longest word.
def Longword(listwords):
    maxlen=0
    for i in listwords:
        if (len(i)>maxlen):
            maxlen=len(i)
    return maxlen
listwords=["I", "am","watching", "a","movie"]
print(Longword(listwords))
```

```python
##### Python function that takes a list of words and return the longest word and the length of the longest word.
def Longword(listwords):
    lst=[]
    for i in listwords:
        lst.append([i,len(i)])
    lst.sort()
    return lst[-1:]
listwords=["I", "am","watching", "a","movie"]
print(Longword(listwords))
```
