```python
print(dir(PatientDB))
```

```python
select_option=(""" Enter : 
- 'insert' to add patient details
- 'View' to Show all patient details
- 'Search' to search for patient details
- 'Disease' to list particular diseased patients
- 'Update' to update patient Status
- 'cured'  to list all cured patients
- 'uncured' to list all patients under treatment
- 'Delete' to delete patient details
- 'Exit' to Exit
Enter Your Choice: 
""")
```

```python
def prompt_insert_details():
    name=input("Enter patient name: ")
    age=int(input("Enter patient age: "))
    disease=input("Enter patient disease: ")
    status=input("Enter cured/uncured: ")
    roomno=int(input("Enter room no: "))
    PatientDB.insert_details(name,age,disease,status,roomno)
```

```python
def view_details():
    Patients=PatientDB.view_all_patients()
    for patient in Patients:
        print(f"Patient Name : {patient['Name']}"
              f"|| Patient Age : {patient['Age']}"
              f"|| Patient Disease : {patient['Disease']}"
              f"|| Patient Status : {patient['Status']}"
              f"|| Patient Room.No : {patient['Roomno']}")
        
```

```python
def prompt_search_patient():
    name=input("Enter Patient name: ")
    patient=PatientDB.search_patient(name)
    if patient:
        print(f"Patient Name : {patient['Name']}"
              f"|| Patient Age : {patient['Age']}"
              f"|| Patient Disease : {patient['Disease']}"
              f"|| Patient Status : {patient['Status']}"
              f"|| Patient Room.No : {patient['Roomno']}")
    else:
        print("Patient Not Found")
        
```

```python
def prompt_diseased_patients():
    disease=input("Enter Disease: ")
    patient=PatientDB.diseased_patients(disease)
    if patient:
        print(f" Diseased Patient Names: {patient}")
    else:
              print(f"No with this {disease} disease")
```

```python
def prompt_update_status():
    name=input("Enter Patient name: ")
    upd_status=input("Enter Status: ")
    PatientDB.update_status(name,upd_status)
```

```python
def prompt_cured_patients():
    patient = PatientDB.cured_patients()
    if patient:
        print("Cured Patients are : ",patient)
```

```python
def prompt_uncured_patients():
    patient = PatientDB.uncured_patients()
    if patient:
        print("Uncured Patients are : ",patient)
```

```python
def delete_patient_details():
    name=input("Enter Patient name: ")
    return PatientDB.delete_patient(name)
```

```python
# - 'insert' to add patient details
# - 'View' to Show all patient details
# - 'Search' to search for patient details
# - 'Disease' to list particular diseased patients
# - 'Update' to update patient Status
# - 'cured'  to list all cured patients
# - 'uncured' to list all patients under treatment
# - 'Delete' to delete patient details
# - 'Exit' to Exit
user_input=input(select_option)
while (user_input != 'Exit'):
    if user_input =='insert':
        prompt_insert_details()
    elif (user_input == 'View'):
        view_details()
    elif (user_input == 'Search'):
        prompt_search_patient()
    elif (user_input =='Disease'):
        prompt_diseased_patients()
    elif (user_input == 'Update'):
        prompt_update_status()
    elif (user_input == 'cured'):
        prompt_cured_patients()
    elif (user_input == 'uncured'):
        prompt_uncured_patients()
    elif (user_input == 'Delete'):
        delete_patient_details()
    else:
        print("Invalid option")
    user_input=input(select_option)
```
