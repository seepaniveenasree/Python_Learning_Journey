### Password Strength Check using Sets

```python
import string
string.punctuation
```

```python
password="@Year2020"
uppcase=set("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
lowcase=set("abcdefghijklmnopqrstuvwxyz")
digits=set("0123456789")
specials=set(string.punctuation)
pwd_chars=set(password)
has_upper= not uppcase.isdisjoint(pwd_chars)
has_lower= not lowcase.isdisjoint(pwd_chars)
has_digit= not digits.isdisjoint(pwd_chars)
has_spc= not specials.isdisjoint(pwd_chars)

if has_upper and has_lower and has_digit and has_spc:
    print("Password is valid")
else:
    print("Password is invalid")
```

```python
#### Remove Stopwords Using Sets
text = "this is a simple text with some common stopwords like and the or in"
stopwords = {"this","is", "a", "with", "some", "like", "and", "the", "or", "in"}
words=text.split()
words
```

```python
filter_words= set(words)-stopwords
print(filter_words)
```

```python
# set1-set2 give unqiue values from set1 and deletes values common on both
# set can perform maths on sets 
filter_words=words-stopwords
print(filter_words)
```

```python
### removing stop word from sentence
text = "this is a simple text with some common stopwords like and the or in"
stopwords = {"this","is", "a", "with", "some", "like", "and", "the", "or", "in"}
lst_words=text.split()
stop_lst=[]
for i in lst_words:
    if i not in stopwords:
        stop_lst.append(i)
print(stop_lst)
```

```python
### Find Pairs with Equal Sum in Two Lists
def find_pairs(lst1,lst2,target):
    return [(x,y) for x in lst1 for y in lst2 if x+y ==target]
print(find_pairs([2,4,3,5],[7,3,4,6,1],8))
```
