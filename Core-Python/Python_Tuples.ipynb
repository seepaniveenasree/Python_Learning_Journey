#### function that takes two tuples and returns a tuple containing the <b>common elements</b> of the input tuples.
t1 = ( 1,2,3,4,5)

t2 = ( 4,5, 6,7,8)

  set : unique values 

  & : intersection , common data 

  set(t1) & set(t2)

  tuple({4,5})

```python
def two_tups(tupp1,tupp2):
    return tuple(set(tupp1) & set(tupp2))
tupp1=tuple(input("Enter tuple1").split())
tupp2=tuple(input("Enter tuple2").split())
print(two_tups(tupp1,tupp2))
```

```python
#general way
def two_tups(tupp1,tupp2):
    com=()
    for i in tupp1:
        if i in tupp2:
            com = com+(i,)
    return com
tupp1=tuple(input("Enter tuple1").split())
tupp2=tuple(input("Enter tuple2").split())
print(two_tups(tupp1,tupp2))
```

```python
### Remove an item from a tuple
def remove_ele(tppl,item):
    lsst=list(tppl)
    if item in lsst:
        lsst.remove(item)
    else:
        print("Item not found")
    tpp1=tuple(lsst)
    return tpp1
tpp1=tuple(input("Enter items: "))
item=input("Enter item to delete: ")
print(remove_ele(tpp1,item))
```

```python
tpp1=tuple(input("Enter items: "))
item=input("Enter item to delete: ")
print(remove_ele(tpp1,item))
```

```python
### Insert a value at a specific position
def insert_ele(tppl,item,pos):
    lsst=list(tppl)
    lsst.insert(pos,item)
    tpp1=tuple(lsst)
    return tpp1
tpp1=tuple(input("Enter items: "))
item=int(input("Enter item to insert: "))
pos=int(input("Enter index: "))
print(insert_ele(tpp1,item,pos))
```

```python
### Find the most frequent element in a tuple
def freq(tpp1):
    freq=0
    res=0
    for i in tpp1:
        if (tpp1.count(i)>freq):
            freq=tpp1.count(i)
            res =i
    return res
tpp1=(1,4,2,6,2,3,2,1,4)
print(freq(tpp1))
```

```python
# way2
### Find the most frequent element in a tuple
def freq(tpp1):
    return max(set(tpp1),key=tpp1.count)
tpp1=(1,4,2,6,2,3,2,1,4)
print(freq(tpp1))
```

```python
### filter()
# filter() is a built-in function used to extract elements from an iterable (like a list, tuple, etc.) based on a condition.
# filter(function, iterable)

#     function → a function that returns True or False
    
#     iterable → sequence (list, tuple, string, etc.)

# It returns a filter object, so you usually convert it to a list.

numb=[1,2,3,4,5]
def div(numb):
    if numb%2==0:
        return True
res=list(filter(div,numb))
print(res)
```

```python
alphabets = ['a', 'b', 'd', 'a', 'e', 'i', 'j', 'o']
def fil_vowels(alphabets):
    vowels=['a','e','i','o','u']
    if (alphabets in vowels):
        return alphabets
unq=list(filter(fil_vowels,alphabets))
print(unq)
```
