```python
#### Function to list of words that are longer than n from a given list of words.
def len_words(sent,length):
    words=sent.split()
    words_match=[]
    for i in words:
        if (len(i)>=length):
            words_match.append(i)
        else:
            continue
    return words_match
sent=input("Enter sentence: ")
length=int(input("Enter length:"))
print(len_words(sent,length))
```

```python
# Right Rotate: lst[-k:] + lst[:-k]
def rotate_list(lst, k):
    n=len(lst)
    k = k % n
    return lst[-k:]+lst[:-k]
nums = [1, 2, 3, 4, 5]
k = 3
result = rotate_list(nums, k)
print(result)
```

```python
# Left Rotate:lst[k:] + lst[:k]
def rotate_list(lst, k):
    n=len(lst)
    k = k % n
    return lst[k:]+lst[:k]
nums = [1, 2, 3, 4, 5]
k = 3
result = rotate_list(nums, k)
print(result)
```

### Case Study: Using Python Lists with Supermarket Data

### Problem Statement

    A supermarket needs a system to:
    
                Store a list of products.
                
                Track the prices of products.
                
                Maintain stock availability.
                
                Allow updates when new stock arrives or when prices change.
                
                Retrieve product details when needed.

| Product | Price | Stock |
| ------- | ----: | ----: |
| Milk    |   2.5 |    10 |
| Eggs    |   3.0 |    20 |
| Bread   |   1.5 |    15 |
| Butter  |   4.0 |     8 |
| Cheese  |   5.0 |     5 |


```python
n=int(input("Enter number of products: "))
products=[]
prices=[]
stocks=[]
for i in range(n):
    product_name,product_price,product_stock=input().split()
    products.append(product_name)
    prices.append(product_price)
    stocks.append(product_stock)
print(products,prices,stocks)
```

```python
##Dislpay avaiable products
def display_products():
    for i in range(len(products)):
        print(f"{products[i]}-{prices[i]}-{stocks[i]}")
display_products()
```

```python
##Updating prices
def upd_price(product_name,newprice):
    if product_name in products:
        pos=products.index(product_name)
        prices[pos]=newprice
        print(f"Updated {product_name} price to $ {newprice}")
    else:
        print("Producct not avaiable")
```

```python
upd_price("Eggs",5.0)
```

```python
display_products()
```

```python
upd_price("Bun",5.0)
```

```python
##Updating stock levels
def upd_stock(product_name,newstock):
    if product_name in products:
        pos=products.index(product_name)
        stocks[pos]=newstock
        print(f"Updated {product_name} stocks to {newstock}")
    else:
        print("Producct not avaiable")
```

```python
upd_stock("Butter",10)
```

```python
# Search for the products
def search_product(productname):
    if productname in products:
        pos=products.index(productname)
        print(f"Product Name: {productname} Product price: {prices[pos]} Product Stock: {stocks[pos]}")
    else:
        print("Product Not found")
```

```python
search_product("Bread")
```

```python
search_product("Drink")
```

```python
# Deleting product
def del_product(productname):
    if productname in products:
        pos=products.index(productname)
        products.remove(productname)
        stocks.pop(pos)
        prices.pop(pos)
        print(f" {productname} is deleted successfully")
    else:
        print("Product not found")
```

```python
del_product("Bread")
```

#  finding pair of numbers
    Start with a list of numbers and a target value.
    
        Take the first number from the list.
    
        Take the next number after it.
    
        Add the two numbers.
    
        Check if their sum equals the target.
    
        If the sum equals the target, print their indexes.

    Repeat this for all possible pairs.

    Stop when all pairs are checked.

```python
def find_pair(lis,target):
    for i in range(len(lis)):
        for j in range(i+1,len(lis)):
            if(lis[i]+lis[j]==target):
                print(i,j)
```

```python
lis=[1,3,4,2,5]
find_pair(lis,5)
```

```python
def find_pair(lis,target):
    pair=[(lis[i],lis[j]) for i in range(len(lis)) for j in range(i+1,len(lis)) if(lis[i]+lis[j]==target)]
    return pair
```

```python
lis=[1,3,4,2,5]
find_pair(lis,5)
```

```python
lis=[1,3,4,2,5]
pairs(lis,5)
```

```python
### find first largest and second-largest numbers without sorting the list
def first_sec_large(lst1):
    first=0
    sec=0
    for i in lst1:
        if(i>first):
            sec=first
            first=i
        elif(i>sec and i<first):
            sec=i
    return first,sec
```

```python
def pair_sum(arr, target_sum):
    result = []
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == target_sum:
                result.append(f"{arr[i]}+{arr[j]}")
    return result

arr = [2, 4, 3, 5, 6, -2, 4, 7, 8, 9]
target_sum = 7
```

```python
print(pair_sum(arr, target_sum))
```
