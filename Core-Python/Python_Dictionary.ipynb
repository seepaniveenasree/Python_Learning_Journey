```python
## Frequency of Character
def char_freq(st1):
    dic={}
    for i in st1:
        if i not in dic.keys():
            dic[i]=1
        else:
            dic[i] +=1
    return dic
st1="Ramayana"
res= char_freq(st1)
print(res)
```

  searching, 
  
  checking, 

  validation, 
    
such as find an item, check whether something exists.


```python
def search_dict(d,value):
    for k,v in d.items():
        if (v==value):
            return k,v
    else:
        return 'Value not found'
d={'Name':'Ram','Age':46,'City':'Ayodhya','Status':'Married','ID':8888,'User':'Ram'}
search_dict(d,8888)
```

```python
# As soon as the first match is found, return terminates the function. 
# The remaining dictionary items are not checked.
search_dict(d,'ram')
```

```python
### If multiple keys can have the same value
def search_dict(d,value):
    lis=[]
    for k,v in d.items():
        if (v==value):
            lis.append((k,v))
    if lis:
        return lis
    else:
        return 'Value not found'
d={'Name':'Ram','Age':46,'City':'Ayodhya','Status':'Married','ID':8888,'User':'Ram'}
search_dict(d,'Ram')
```

```python
# append() stores every matching pair. 
# The function returns the list only after the loop finishes.
```

```python
def searchDictAll(d,value):
    res=[(k,v) for k,v in d.items() if v==value]
    return res
```

```python
d={'Name':'Ram','Age':46,'City':'Ayodhya','Status':'Married','ID':8888,'User':'Ram'}
searchDictAll(d,'Ram')
```

### Dict comprehension

```python
# Define a function which takes as a parameter dictionary and returns a dictionary in which reverse the key-value pairs are reversed.

dicc={'A':20,'B':'bb','C':40,'D':'aa'}
def rev_dic(dicc):
    reverse={v:k for k,v in dicc.items()}
    return reverse
rev_dic(dicc)
```

```python
# Define a function which takes as a parameter dictionary and returns a dictionary in which everse the key-value pairs are reversed.

my_dict = {'a': 1, 'b': 2, 'c': 3}

def reverse_dict(my_dict):
    return {v: k for k, v in my_dict.items()}
reverse_dict(my_dict)
```

```python
# Define a function that takes a dictionary as a parameter and returns a dictionary with elements based on a condition.

my_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4} 

def filter_dict(my_dict, condition):

    return {k: v for k, v in my_dict.items() if condition(k, v)}
```

```python

filtered_dict = filter_dict(my_dict, lambda k, v: v % 2 == 0) 

print(filtered_dict)
```

### Dictionary Comprehension
Dictionary comprehension in Python allows you to construct a dictionary in a single line using a concise and readable syntax

         {key: value for item in iterable if condition}

```python
sqr_dic={i:i**2 for i in range(1,6)}
print(sqr_dic)
```

```python
sqr_dic1={i:i**2 for i in range(1,20) if (i%2==0)}
print(sqr_dic1)
```

```python
dict1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
twice={k:v*2 for k,v in dict1.items()}
print(twice)
```

```python
# Convert to range of numbers into Capital letters 
dic={i:chr(97+i) for i in range(11)}
print(dic)
```

```python
dic={i:chr(65+i) for i in range(11)}
print(dic)
```

```python
#### list elements as the keys and the length of each string as the values
dict_comp = ["allow","sequences","to","be","built","from","other","sequences"]
len_comp ={ k:len(k) for k in dict_comp}
print(len_comp)
```

```python
dict_comp = ["allow","sequences","to","be","built","from","other","sequences"]
upp_comp ={ k:k.upper() for k in dict_comp}
print(upp_comp)
```

```python
dict_comp = ["allow","sequences","to","be","built","from","other","sequences"]
low_comp ={ k:k.lower() for k in dict_comp}
print(low_comp)
```

```python
dict_comp = ["allow","sequences","to","be","built","from","other","sequences"]
cap_comp ={ k:k.capitalize() for k in dict_comp}
print(cap_comp)
```
