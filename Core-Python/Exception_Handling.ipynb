## Exception handling is used

    * To handle runtime errors in a Python program without terminating the    
    program abnormally.

    * An exception is an error that occurs during program execution.


Python provides 5 important keywords for exception handling.

Keyword           	Purpose
------------------------------------------------------------------------

try	:-                Defines a block of code where an exception may occur.
    
except	:-           Handles an exception.

else	:-           Executes when no exception occurs in the try block.

finally	  :-         Executes whether an exception occurs or not.

raise	:-           Manually raises an exception.

### If we already have if-else, why do we need exception handling?

The main difference is:
        
    - if-else handles conditions that we can check in advance.
      
    - Exception handling handles unexpected runtime errors that interrupt normal program execution.

if-else	                       |                     Exception handling
--------------------------------------------------------

Checks conditions	       |               Handles exceptions
 
Used for expected decisions	|              Used for runtime errors

Does not require an error	|              Executes when an exception occurs

Uses if, elif, else	        |              Uses try, except, else, finally

## finally
finally is optional

Finally is a block in exception that always execute,regardless of wheather an exception occurs or not

```python
try:
    a=int(input("Enter first number:"))
    b=int(input("Enter second number:"))
    result=a/b
    print("Result: ",result)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Division successfull")
print("End")
```

```python
try:
    a=int(input("Enter first number:"))
    b=int(input("Enter second number:"))
    result=a/b
    print("Result: ",result)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Division successfull")
print("End")
```

### Manually raising exceptions

    The raise keyword is used to manually generate an exception when a particular condition is not satisfied.

    raise ExceptionType("Error message")

```python
# Age Validation
try:
    age=int(input("Enter your age:"))
    if age<18:
        raise ValueError("You cannot vote in elections")
    if age>=18:
        raise Exception("U are eligible to vote")
    print("Eligible to vote")
except ValueError as e:
    print("Value Error: ",e)
except Exception as e:
    print("Error: ",e)
```
